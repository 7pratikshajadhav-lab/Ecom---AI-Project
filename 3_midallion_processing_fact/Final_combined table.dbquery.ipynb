```sql
CREATE OR REPLACE VIEW ecommerce.gold.fact_transaction_denorm AS(
    SELECT i.*,
    c.year,
    c.manth_name,
    c.day_name,
    c.is_weekend,
    c.quarter,
    c.week,
    p.sku,
    p.category_code,
    p.category_name,
    p.brand_code,
    p.brand_name,
    p.color,
    p.size,
    p.rating_count,
    extract(HOUR FROM i.transaction_date) as hour_of_day
    FROM ecommerce.gold.gld_fact_order_items as i 
    JOIN ecommerce.gold.gld_dim_date as c ON i.date_id=c.date_id
    JOIN ecommerce.gold.gld_dim_products as p ON i.product_id=p.product_id
    
);
```
