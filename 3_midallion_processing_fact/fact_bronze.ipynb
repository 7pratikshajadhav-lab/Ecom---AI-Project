```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, DateType, TimestampType
import pyspark.sql.functions as F
```

```python
catalog_name="ecommerce"
```

```python
order_item_schema = StructType([
    StructField("dt", StringType(), True),
    StructField("order_ts", StringType(), True),
    StructField("cutomer_id", StringType(), True),
    StructField("order_id", StringType(), True),
    StructField("item_seq", StringType(), True),
    StructField("product_id", StringType(), True),
    StructField("quantity", StringType(), True),
    StructField("unit_price_currency", StringType(), True),
    StructField("unit_price", StringType(), True),
    StructField("discount_pct", StringType(), True),
    StructField("tax_amount", StringType(), True),
    StructField("channel", StringType(), True),
    StructField("coupon_code", StringType(), True),
    StructField("transaction_ts",StringType(), True)
])
```

```python
row_date_path="/Volumes/ecommerce/source_data/row/order_items/landing/order_items*.csv"
df= spark.read.option("header","true").option("delimeter",",").schema(order_item_schema).csv(row_date_path)\
    .withColumn("file_name", F.col("_metadata.file_path"))\
    .withColumn("ingest_timestamp", F.current_timestamp())
```

```python
display(df)
```

```python
df.write.format("delta")\
    .mode("overwrite")\
    .option("mergeSchema","true")\
    .saveAsTable(f"{catalog_name}.bronze.brz_order_item")
```
